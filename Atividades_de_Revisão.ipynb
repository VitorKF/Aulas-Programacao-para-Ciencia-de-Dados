**Atividade I**

Escreva um programa em Python que gere um número aleatório de 0 a 15, usando a biblioteca 'random' e o método 'randint()'. Na sequência, utilizando um loop faça com que o usuário tente acertar o número gerado.

A cada vez que o número é acertado, armazene num arquivo o número gerado e a quantidade de tentativas que o usuário realizou para acertar. Quando o usuário decidir parar de adivinhar, mostre todo o histórico que ele construiu durante a execução do programa.

```python
import random

registros = []


def adivinharNumero():
    numero = random.randint(0, 15)
    tentativas = []

    while True:
        adivinha = int(input("Tente adivinhar o número: "))
        tentativas.append(adivinha)

        if adivinha == numero:
            print("\nParabéns você acertou o número {}".format(numero))

            registros.append(tentativas)

            with open('registro.txt', 'w') as arquivo:
                arquivo.write(str(registros))

            break
        else:
            print("Tente Novamente")


def verHistorico():
    if len(registros) == 0:
        print("\nNenhum número no registro ainda")
    else:
        print("\nO usuário jogou {} vezes.".format(len(registros)))

        for numeroJogo in range(len(registros)):
            tentativas = registros[numeroJogo]

            print("\nPara o jogo {}:".format(numeroJogo + 1))
            print("O usuário demorou {} chances.".format(len(tentativas)))

            for tentativa in range(len(tentativas)):
                print("Tentativa número {} = {}".format(
                    tentativa + 1,
                    tentativas[tentativa]
                ))


while True:
    print("\n================================")
    print("       ADIVINHE O NÚMERO")
    print("================================")
    print("1 - Adivinhar um número")
    print("2 - Ver histórico")
    print("3 - Sair")
    print("================================")

    opcao = input("Digite uma opção: ")

    if opcao == "1":
        adivinharNumero()

    elif opcao == "2":
        verHistorico()

    elif opcao == "3":
        print("\nPrograma encerrado")
        break

    else:
        print("\nErro! Digite um número de 1 a 3")
```

**Atividade II**

Crie um programa que permita adicionar itens para uma lista de compras.

O usuário vai adicionando a quantidade e o valor de cada item da lista.

Ao final, deve mostrar o valor total das compras e salvar a lista num arquivo .CSV.

```python
import csv

produtos = []


def cadastrarProduto():
    nome = input("\nDigite o nome do produto: ")
    quantidade = int(input("Digite a quantidade: "))
    preco = float(input("Digite o valor do produto: "))

    produto = {
        "nome": nome,
        "quantidade": quantidade,
        "preco": preco
    }

    produtos.append(produto)

    print("\nProduto adicionado com sucesso!")


def consultarProdutos():
    if len(produtos) == 0:
        print("\nNenhum produto na lista de compras.")
    else:
        total = 0

        print("\nLista de compras:")

        for produto in produtos:
            valor = produto["quantidade"] * produto["preco"]
            total = total + valor

            print("\nProduto: {}".format(produto["nome"]))
            print("Quantidade: {}".format(produto["quantidade"]))
            print("Valor unitário: R$ {:.2f}".format(produto["preco"]))
            print("Valor total: R$ {:.2f}".format(valor))

        print("\nValor total das compras: R$ {:.2f}".format(total))


def salvarCSV():
    with open("lista_compras.csv", "w", newline="", encoding="utf-8") as arquivo:
        escritor = csv.writer(arquivo)

        escritor.writerow(["Produto", "Quantidade", "Valor Unitário", "Valor Total"])

        for produto in produtos:
            valor = produto["quantidade"] * produto["preco"]

            escritor.writerow([
                produto["nome"],
                produto["quantidade"],
                produto["preco"],
                valor
            ])

        print("\nLista salva no arquivo lista_compras.csv")


while True:
    print("\n================================")
    print("       LISTA DE COMPRAS")
    print("================================")
    print("1 - Adicionar um produto")
    print("2 - Ver lista de compras")
    print("3 - Sair")
    print("================================")

    opcao = input("Digite uma opção: ")

    if opcao == "1":
        cadastrarProduto()

    elif opcao == "2":
        consultarProdutos()

    elif opcao == "3":
        salvarCSV()
        print("\nPrograma encerrado")
        break

    else:
        print("\nErro! Digite um número de 1 a 3")
```
