Construa uma calculadora básica usando os conceitos de orientação a objetos.
Use a função raiz quadrada usando o método de Newton (se achar complicado use a função SQRT).

```python
import math


class Calculadora:

    def soma(self, a, b):
        return a + b

    def subtracao(self, a, b):
        return a - b

    def multiplicacao(self, a, b):
        return a * b

    def divisao(self, a, b):
        if b == 0:
            print("Não é possível dividir por zero.")
            return None

        return a / b

    def potenciacao(self, a, b):
        return a ** b

    def radiciacao(self, a):
        if a < 0:
            print("Não é possível calcular a raiz de um número negativo.")
            return None

        return math.sqrt(a)


def iniciarCalculadora():

    calculadora = Calculadora()

    while True:
        print("\n======================================================")
        print("       ESCOLHA QUAL OPERAÇÃO VOCÊ DESEJA FAZER")
        print("======================================================")
        print("1 - SOMA")
        print("2 - SUBTRAÇÃO")
        print("3 - MULTIPLICAÇÃO")
        print("4 - DIVISÃO")
        print("5 - POTENCIAÇÃO")
        print("6 - RADICIAÇÃO")
        print("7 - SAIR")
        print("======================================================")

        opcao = input("Digite uma opção: ")

        if opcao == "1":
            a = float(input("Digite o primeiro número: "))
            b = float(input("Digite o segundo número: "))

            print("Resultado: {}".format(calculadora.soma(a, b)))

        elif opcao == "2":
            a = float(input("Digite o primeiro número: "))
            b = float(input("Digite o segundo número: "))

            print("Resultado: {}".format(calculadora.subtracao(a, b)))

        elif opcao == "3":
            a = float(input("Digite o primeiro número: "))
            b = float(input("Digite o segundo número: "))

            print("Resultado: {}".format(calculadora.multiplicacao(a, b)))

        elif opcao == "4":
            a = float(input("Digite o primeiro número: "))
            b = float(input("Digite o segundo número: "))

            resultado = calculadora.divisao(a, b)

            if resultado is not None:
                print("Resultado: {}".format(resultado))

        elif opcao == "5":
            a = float(input("Digite a base: "))
            b = float(input("Digite o expoente: "))

            print("Resultado: {}".format(calculadora.potenciacao(a, b)))

        elif opcao == "6":
            a = float(input("Digite o número: "))

            resultado = calculadora.radiciacao(a)

            if resultado is not None:
                print("Resultado: {}".format(resultado))

        elif opcao == "7":
            print("\nPrograma encerrado")
            break

        else:
            print("\nErro! Digite um número de 1 a 7")


iniciarCalculadora()
```
