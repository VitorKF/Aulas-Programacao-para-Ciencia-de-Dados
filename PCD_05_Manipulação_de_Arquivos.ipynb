Atividade I

Escreva um programa que receba do usuário os valores de um conjunto de características de um personagem inventado por você.
Leia os valores e grave um arquivo para posterior manipulação.
Escolha o formato de gravação.

O arquivo não precisa ser alterado, apenas sobreescrito.
Crie uma função para ler e apresentar (print) os dados gravados.

```python
import json

personagem = '{"nome": "Felipe", "habilidade": "Quebrar Códigos"}'

personagem_dic = json.loads(personagem)

with open('sample_data/personagem.txt', 'w') as json_file:
  json.dump(personagem_dic, json_file, indent = 4)

print(personagem_dic)

personagem_dic['nome'] = str(input("Qual o nome de seu personagem: "))
personagem_dic['habilidade'] = str(input("Qual a habilidade de seu personagem: "))
with open('sample_data/personagem.txt', 'w') as json_file:
  json.dump(personagem_dic, json_file, indent = 4)

def ler_personagem():
    with open('sample_data/personagem.txt', 'r') as json_file:
        personagem_dic = json.load(json_file)

    print(personagem_dic)

ler_personagem()
```

Atividade II

Gere uma lista com 20 números aleatórios e, para cada valor, calcule o resultado da multiplicação por 2 e por 10. Salve os dados em TXT, JSON, e CSV contendo três campos: valor original, valor *2, valor *10.

Em ambas as atividades
Ao final , adicione os arquivos gerados ao repositório do GitHub e realize o comit e push das alterações.

```python
import random
import json
import csv

numeros = []

for i in range(20):
    numeros.append(random.randint(1, 100))

dados = []

for numero in numeros:
    dados.append({
        "valor original": numero,
        "valor *2": numero * 2,
        "valor *10": numero * 10
    })

with open('sample_data/numeros.txt', 'w') as arquivo:
    for dado in dados:
        arquivo.write('{};{};{}\n'.format(
            dado["valor original"],
            dado["valor *2"],
            dado["valor *10"]
        ))

with open('sample_data/numeros.json', 'w') as arquivo:
    json.dump(dados, arquivo, indent=4)

with open('sample_data/numeros.csv', 'w', newline='') as arquivo:
    escritor = csv.writer(arquivo)

    escritor.writerow(["valor original", "valor *2", "valor *10"])

    for dado in dados:
        escritor.writerow([
            dado["valor original"],
            dado["valor *2"],
            dado["valor *10"]
        ])

print("Arquivos gerados com sucesso!")
```
