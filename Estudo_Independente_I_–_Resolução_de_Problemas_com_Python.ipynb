Desafio 2 – Cadastro e consulta de produtos
Desenvolva um programa que permita cadastrar produtos utilizando um dicionário.
Cada produto deverá possuir:
* código;
* nome;
* preço;
* quantidade em estoque.
Ao final, o programa deverá permitir:
* consultar um produto pelo código;
* calcular o valor total do estoque;
* identificar o produto de maior valor unitário.

Organize a solução utilizando funções.

```python
def cadastrarProdutos():
    produtos = {}

    quantidade = int(input("Digite a quantidade de produtos: "))

    for i in range(quantidade):
        print("\nProduto {}".format(i + 1))

        codigo = input("Digite o código: ")
        nome = input("Digite o nome: ")
        preco = float(input("Digite o preço: "))
        estoque = int(input("Digite a quantidade em estoque: "))

        produtos[codigo] = {
            "nome": nome,
            "preco": preco,
            "estoque": estoque
        }

    return produtos

def consultarProduto(produtos):
    codigo = input("\nDigite o código do produto que deseja consultar: ")

    if codigo in produtos:
        print("\nProduto encontrado:")
        print("Nome: {}".format(produtos[codigo]["nome"]))
        print("Preço: R$ {:.2f}".format(produtos[codigo]["preco"]))
        print("Quantidade em estoque: {}".format(produtos[codigo]["estoque"]))
    else:
        print("Produto não encontrado.")

def valorTotalEstoque(produtos):
    total = 0

    for produto in produtos.values():
        total = total + produto["preco"] * produto["estoque"]

    return total

def maiorValorUnitario(produtos):
    maior = None

    for codigo, produto in produtos.items():
        if maior is None or produto["preco"] > maior["preco"]:
            maior = produto
            maior["codigo"] = codigo

    return maior

produtos = cadastrarProdutos()

while True:

    codigo = input("\nDigite o código do produto que deseja consultar ou SAIR para sair: ")

    if codigo == "SAIR":
        break

    if codigo in produtos:
        print("\nProduto encontrado:")
        print("Nome: {}".format(produtos[codigo]["nome"]))
        print("Preço: R$ {:.2f}".format(produtos[codigo]["preco"]))
        print("Quantidade em estoque: {}".format(produtos[codigo]["estoque"]))
    else:
        print("Produto não encontrado.")


total = valorTotalEstoque(produtos)

print("\nValor total do estoque: R$ {:.2f}".format(total))

maior = maiorValorUnitario(produtos)

print("\nProduto de maior valor unitário:")
print("Código: {}".format(maior["codigo"]))
print("Nome: {}".format(maior["nome"]))
print("Preço: R$ {:.2f}".format(maior["preco"]))
```


Desafio 3 – Sistema de Estatísticas
Desenvolva um programa que leia uma sequência de números informados pelo usuário até que seja digitado
um valor de encerramento.
Ao final, o programa deverá apresentar:
* quantidade de valores informados;
* média;
* mediana;
* maior valor;
* menor valor;
* quantidade de números pares;
* quantidade de números ímpares.

Organize todos os cálculos em funções independentes.

```python
def calcularMedia(numeros):
    soma = 0

    for numero in numeros:
        soma = soma + numero

    return soma / len(numeros)


def calcularMediana(numeros):
    numerosOrdenados = sorted(numeros)

    quantidade = len(numerosOrdenados)
    meio = quantidade // 2

    if quantidade % 2 == 0:
        mediana = (numerosOrdenados[meio - 1] + numerosOrdenados[meio]) / 2
    else:
        mediana = numerosOrdenados[meio]

    return mediana


def maiorValor(numeros):
    maior = numeros[0]

    for numero in numeros:
        if numero > maior:
            maior = numero

    return maior


def menorValor(numeros):
    menor = numeros[0]

    for numero in numeros:
        if numero < menor:
            menor = numero

    return menor


def quantidadePares(numeros):
    pares = 0

    for numero in numeros:
        if numero % 2 == 0:
            pares = pares + 1

    return pares


def quantidadeImpares(numeros):
    impares = 0

    for numero in numeros:
        if numero % 2 != 0:
            impares = impares + 1

    return impares


numeros = []

while True:
    entrada = input("Digite um número ou SAIR para encerrar: ")

    if entrada == "SAIR":
        break

    try:
        numero = int(entrada)
        numeros.append(numero)
    except ValueError:
        print("Valor inválido. O valor não foi contado.")


if len(numeros) > 0:
    print("\nQuantidade de valores informados: {}".format(len(numeros)))
    print("Média: {:.2f}".format(calcularMedia(numeros)))
    print("Mediana: {:.2f}".format(calcularMediana(numeros)))
    print("Maior valor: {}".format(maiorValor(numeros)))
    print("Menor valor: {}".format(menorValor(numeros)))
    print("Quantidade de números pares: {}".format(quantidadePares(numeros)))
    print("Quantidade de números ímpares: {}".format(quantidadeImpares(numeros)))
else:
    print("\nNenhum número foi informado.")
```
