Uma empresa de comércio eletrônico deseja analisar o desempenho de suas vendas durante uma semana.
Foram registrados, para cada dia, os seguintes indicadores:
* quantidade de pedidos;
* faturamento (R$);
* número de clientes;
* valor médio dos pedidos;
* quantidade de reclamações

```python
# Importar a biblioteca NumPy
import numpy as np

# Criar o array com os dados de vendas
# Colunas: Pedidos, Faturamento, Clientes, Ticket Médio, Reclamações
dados = np.array([
    [120, 15000, 100, 125.00, 4],
    [135, 16200, 108, 120.00, 5],
    [98, 11700, 82, 119.39, 3],
    [150, 21000, 125, 140.00, 7],
    [175, 24500, 140, 140.00, 9],
    [160, 22400, 132, 140.00, 6],
    [110, 13200, 95, 120.00, 2]
])

# Criar o array com os dias da semana
dias = np.array([
    "Segunda",
    "Terça",
    "Quarta",
    "Quinta",
    "Sexta",
    "Sábado",
    "Domingo"
])
```

```python
# 1.1 - Características do array
print("Quantidade de linhas:", dados.shape[0])
print("Quantidade de colunas:", dados.shape[1])
print("Quantidade total de valores:", dados.size)
print("Número de dimensões:", dados.ndim)
print("Tipo dos dados:", dados.dtype)
```

```python
# 2.1 - Dados referentes à sexta-feira
print("Dados de sexta-feira:")
print(dados[4])

# 2.2 - Faturamento de todos os dias
print("\nFaturamento:")
print(dados[:, 1])

# 2.3 - Quantidade de reclamações
print("\nReclamações:")
print(dados[:, 4])

# 2.4 - Dados de quinta-feira até domingo
print("\nQuinta-feira até domingo:")
print(dados[3:7])

# 2.5 - Colunas de pedidos, faturamento e clientes
print("\nPedidos, faturamento e clientes:")
print(dados[:, 0:3])
```

```python
# 3.1 - Total de pedidos
total_pedidos = np.sum(dados[:, 0])

# 3.2 - Faturamento total
faturamento_total = np.sum(dados[:, 1])

# 3.3 - Quantidade total de clientes
total_clientes = np.sum(dados[:, 2])

# 3.4 - Média diária de pedidos
media_pedidos = np.mean(dados[:, 0])

# 3.5 - Média diária de faturamento
media_faturamento = np.mean(dados[:, 1])

# 3.6 - Média de reclamações por dia
media_reclamacoes = np.mean(dados[:, 4])

# Exibir resultados
print("Total de pedidos:", total_pedidos)
print("Faturamento total: R$", faturamento_total)
print("Total de clientes:", total_clientes)
print("Média diária de pedidos:", round(media_pedidos, 2))
print("Média diária de faturamento: R$", round(media_faturamento, 2))
print("Média de reclamações:", round(media_reclamacoes, 2))
```

```python
# 4.1 - Maior e menor quantidade de pedidos
indice_maior_pedidos = np.argmax(dados[:, 0])
indice_menor_pedidos = np.argmin(dados[:, 0])

print("Maior quantidade de pedidos:", dias[indice_maior_pedidos])
print("Menor quantidade de pedidos:", dias[indice_menor_pedidos])

# 4.2 - Maior e menor faturamento
indice_maior_faturamento = np.argmax(dados[:, 1])
indice_menor_faturamento = np.argmin(dados[:, 1])

print("\nMaior faturamento:", dias[indice_maior_faturamento])
print("Menor faturamento:", dias[indice_menor_faturamento])

# 4.3 - Amplitude do faturamento
maior_faturamento = np.max(dados[:, 1])
menor_faturamento = np.min(dados[:, 1])

amplitude = maior_faturamento - menor_faturamento

print("\nAmplitude do faturamento: R$", amplitude)
```

```python
# 5.1 - Dias com faturamento superior a R$ 12.000
filtro_faturamento = dados[:, 1] > 12000
print("Dias com faturamento acima de R$ 12.000:")
print(dias[filtro_faturamento])

# 5.2 - Dias com mais de 150 pedidos
filtro_pedidos = dados[:, 0] > 150
print("\nDias com mais de 150 pedidos:")
print(dias[filtro_pedidos])

# 5.3 - Dias com mais de 4 reclamações
filtro_reclamacoes = dados[:, 4] > 4
print("\nDias com mais de 4 reclamações:")
print(dias[filtro_reclamacoes])

# 5.4 - Faturamento dos dias com mais de 150 pedidos
print("\nFaturamento dos dias com mais de 150 pedidos:")
print(dados[filtro_pedidos, 1])

# 5.5 - Quantidade de dias com faturamento abaixo da média
filtro_abaixo_media = dados[:, 1] < np.mean(dados[:, 1])
quantidade_dias = np.count_nonzero(filtro_abaixo_media)

print("\nDias com faturamento abaixo da média:", quantidade_dias)
print(dias[filtro_abaixo_media])
```

```python
# Classificar os dias conforme o faturamento
classificacao = np.where(
    dados[:, 1] >= 20000,
    "Alto desempenho",
    "Desempenho normal"
)

# 6.1 - Apresentar a classificação de cada dia
print("Classificação dos dias:")
print(np.column_stack((dias, classificacao)))

# 6.2 - Contar os dias de alto desempenho
quantidade_alto_desempenho = np.count_nonzero(
    classificacao == "Alto desempenho"
)

print("\nDias classificados como alto desempenho:")
print(quantidade_alto_desempenho)
```

```python
# Separar os valores de faturamento
faturamento = dados[:, 1]

# Calcular as medidas estatísticas
media = np.mean(faturamento)
mediana = np.median(faturamento)
desvio_padrao = np.std(faturamento)
minimo = np.min(faturamento)
maximo = np.max(faturamento)

# Exibir resultados
print("Média: R$", round(media, 2))
print("Mediana: R$", round(mediana, 2))
print("Desvio padrão: R$", round(desvio_padrao, 2))
print("Mínimo: R$", minimo)
print("Máximo: R$", maximo)
```

```python
# 8.1 - Média de cada indicador
medias = np.mean(dados, axis=0)

# 8.2 - Maior valor de cada indicador
maiores_valores = np.max(dados, axis=0)

# 8.3 - Menor valor de cada indicador
menores_valores = np.min(dados, axis=0)

# 8.4 - Coeficiente de variação de cada indicador
desvios = np.std(dados, axis=0)
variacao_relativa = (desvios / medias) * 100

# Exibir resultados
indicadores = np.array([
    "Pedidos",
    "Faturamento",
    "Clientes",
    "Ticket Médio",
    "Reclamações"
])

print("Média de cada indicador:")
print(np.column_stack((indicadores, medias.round(2))))

print("\nMaiores valores:")
print(np.column_stack((indicadores, maiores_valores)))

print("\nMenores valores:")
print(np.column_stack((indicadores, menores_valores)))

print("\nVariação relativa (%):")
print(np.column_stack((indicadores, variacao_relativa.round(2))))

# Identificar o indicador com maior variação relativa
indice_maior_variacao = np.argmax(variacao_relativa)

print("\nIndicador com maior variação relativa:")
print(indicadores[indice_maior_variacao])
```

```python
# 9.1 - Melhor dia considerando o faturamento
melhor_dia = dias[np.argmax(dados[:, 1])]

print("Melhor dia considerando o faturamento:", melhor_dia)

# 9.2 - Verificar se o melhor dia também teve mais pedidos
dia_maior_pedido = dias[np.argmax(dados[:, 0])]

print("Dia com maior quantidade de pedidos:", dia_maior_pedido)
print("É o mesmo dia?", melhor_dia == dia_maior_pedido)

# 9.3 - Verificar se o dia com maior faturamento teve mais reclamações
dia_maior_reclamacao = dias[np.argmax(dados[:, 4])]

print("\nDia com maior faturamento:", melhor_dia)
print("Dia com mais reclamações:", dia_maior_reclamacao)
print("É o mesmo dia?", melhor_dia == dia_maior_reclamacao)

# 9.4 - Dias com mais de 150 pedidos e faturamento acima de R$ 20.000
filtro_integrado = (
    (dados[:, 0] > 150) &
    (dados[:, 1] > 20000)
)

quantidade_dias_integrados = np.count_nonzero(filtro_integrado)

print("\nQuantidade de dias que atendem às duas condições:")
print(quantidade_dias_integrados)

print("Dias encontrados:")
print(dias[filtro_integrado])
```

```python
# RESUMO DA SEMANA
# Identificar o melhor e o pior dia pelo faturamento
melhor_dia = dias[np.argmax(dados[:, 1])]
pior_dia = dias[np.argmin(dados[:, 1])]

# Identificar o maior número de reclamações
maior_reclamacao = np.max(dados[:, 4])

# Identificar os dias de alto desempenho
dias_alto_desempenho = dias[dados[:, 1] >= 20000]

# Exibir o resumo
print("=" * 40)
print("          RESUMO DA SEMANA")
print("=" * 40)

print("Total de pedidos:", np.sum(dados[:, 0]))
print("Faturamento total: R$", round(np.sum(dados[:, 1]), 2))
print("Média diária de pedidos:", round(np.mean(dados[:, 0]), 2))
print("Média diária de faturamento: R$", round(np.mean(dados[:, 1]), 2))
print("Melhor dia:", melhor_dia)
print("Pior dia:", pior_dia)
print("Maior número de reclamações:", maior_reclamacao)

print("Dias de alto desempenho:")
print(dias_alto_desempenho)

print("=" * 40)
```
