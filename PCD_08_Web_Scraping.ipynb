```python
import requests
from bs4 import BeautifulSoup
```

```python
url = (
    "https://store.steampowered.com/search/?specials=1&cc=br&l=portuguese"
)
pagina = requests.get(url)
print("Status:", pagina.status_code)
```

```python
soup = BeautifulSoup(
  pagina.text,
  "html.parser"
)
```

```python
produtos = soup.select(
  "a.search_result_row"
)
```

```python
for produto in produtos[:5]:
  nome = produto.select_one(
    "span.title"
  ).get_text(strip=True)
  preco_original = produto.select_one(
    ".discount_original_price"
  ).get_text(strip=True)
  preco_promocional = produto.select_one(
    ".discount_final_price"
  ).get_text(strip=True)
  desconto = produto.select_one(
    ".discount_pct"
  ).get_text(strip=True)
  print("Nome:", nome)
  print("Preço original:", preco_original)
  print("Desconto:", desconto)
  print("Preço promocional:", preco_promocional)
  print("-" * 50)
```

```python
import google_colab_selenium as gs
from selenium.webdriver.common.by import By
from time import sleep
```

```python
navegador = gs.Chrome()
```

```python
url = (
  "https://www.ecb.europa.eu/stats/policy_and_exchange_rates/euro_reference_exchange_rates/html/index.en.html"
)
```

```python
navegador.get(url)
sleep(2)
print("Página:", navegador.title)
```

```python
botoes = navegador.find_elements(
  By.XPATH,
  "//*[@id='cookieConsent']//button"
  "[contains(., 'I do not accept')]"
)
```

```python
for botao in botoes:
  if botao.is_displayed():
    navegador.execute_script(
      "arguments[0].click();", botao
    )
  break
sleep(1)
```

```python
dolar = navegador.find_element(
  By.XPATH,
  "//tr[td[normalize-space()='USD']]/td[3]"
).text
```

```python
real = navegador.find_element(
  By.XPATH,
  "//tr[td[normalize-space()='BRL']]/td[3]"
).text
```

```python
print("\nCOTAÇÕES:")
print("1 EUR =", dolar, "USD")
print("1 EUR =", real, "BRL")
# Conversão dólar -> real
dolar_real = float(real) / float(dolar)
print("1 USD =", round(dolar_real, 2), "BRL")
```

```python
link = navegador.find_element(
  By.XPATH,
  "//tr[td[normalize-space()='USD']]//a"
)
link.click()
sleep(2)
```

```python
print("\nPágina após o clique:")
print(navegador.title)
print(navegador.current_url)
```

```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.common.keys import Keys
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
import time
```

```python
# Configurar o navegador
opcoes = webdriver.ChromeOptions()

opcoes.binary_location = "/usr/bin/google-chrome"

opcoes.add_argument("--headless=new")
opcoes.add_argument("--no-sandbox")
opcoes.add_argument("--disable-dev-shm-usage")
opcoes.add_argument("--disable-gpu")

# Iniciar o navegador
navegador = webdriver.Chrome(options=opcoes)

# Definir tamanho da janela
navegador.set_window_size(1920, 1080)

print("Navegador iniciado com sucesso!")

navegador.get("https://www.climatempo.com.br/")

WebDriverWait(navegador, 20).until(
    lambda driver: driver.execute_script(
        "return document.readyState"
    ) == "complete"
)

print("Site acessado com sucesso!")

# Solicitar o nome da cidade
cidade = input("Digite o nome da cidade: ")

# Aguardar o campo de pesquisa ficar disponível
campo_busca = WebDriverWait(navegador, 20).until(
    EC.element_to_be_clickable((By.ID, "searchGeneral"))
)

# Digitar o nome da cidade
campo_busca.clear()
campo_busca.send_keys(cidade)

# Aguardar as sugestões
time.sleep(3)

# Localizar somente os links das sugestões
sugestoes = navegador.find_elements(
    By.CSS_SELECTOR,
    'a[data-id^="Menu_MainSearch_Item_NavLink"]'
)

# Armazenar as cidades encontradas
opcoes_validas = []

for sugestao in sugestoes:
    if sugestao.is_displayed():
        texto = sugestao.text.strip()

        if texto:
            opcoes_validas.append({
                "nome": texto,
                "link": sugestao.get_attribute("href")
            })

# Exibir as opções
print("\nCidades encontradas:")

for i, opcao in enumerate(opcoes_validas, start=1):
    print(f"{i} - {opcao['nome']}")

    # Solicitar a escolha do usuário
escolha = int(input("\nEscolha o número da cidade: "))

# Verificar se a opção é válida
if escolha < 1 or escolha > len(opcoes_validas):
    print("Opção inválida!")

else:
    # Obter a cidade selecionada
    cidade_escolhida = opcoes_validas[escolha - 1]

    # Acessar a página da cidade
    navegador.get(cidade_escolhida["link"])

    # Aguardar o carregamento das temperaturas
    WebDriverWait(navegador, 20).until(
        EC.presence_of_element_located((By.ID, "min-temp-1"))
    )

    # Extrair a temperatura mínima
    temperatura_minima = navegador.find_element(
        By.ID, "min-temp-1"
    ).get_attribute("textContent")

    # Extrair a temperatura máxima
    temperatura_maxima = navegador.find_element(
        By.ID, "max-temp-1"
    ).get_attribute("textContent")

    # Exibir os resultados
    print("\n--- PREVISÃO DO TEMPO ---")
    print("Cidade:", cidade_escolhida["nome"])
    print("Temperatura mínima:", temperatura_minima)
    print("Temperatura máxima:", temperatura_maxima)

# Encerrar o navegador
navegador.quit()

print("\nPrograma finalizado!")
```
