Usandos os métodos do Pandas, localize a sua cidade de residência e:
- mostre todas as informações
- qual o mês de temperatura mínima
- qual o mês de maior déficit hídrico
- a média de temperatura pelas 4 estações do ano

```python
import pandas as pd
import numpy as np

# Importando a base de dados dos municípios
df = pd.read_csv(
    'https://raw.githubusercontent.com/andr3nf/clima_municipios_brasileiros/refs/heads/main/clima_historico_municipal.csv'
)

# Criando uma cópia do DataFrame original
df_santa_rosa = df.copy()

# Localizando Santa Rosa - RS
santa_rosa = df_santa_rosa.loc[
    (df_santa_rosa['nome_municipio'] == 'Santa Rosa') &
    (df_santa_rosa['uf'] == 'RS')
]

# Exibindo todas as informações do município
display(santa_rosa)
```

```python
# Selecionando as colunas de temperatura mínima
colunas_tmin = [
    'tmin_jan', 'tmin_fev', 'tmin_mar', 'tmin_abr',
    'tmin_mai', 'tmin_jun', 'tmin_jul', 'tmin_ago',
    'tmin_set', 'tmin_out', 'tmin_nov', 'tmin_dez'
]

# Obtendo os valores de temperatura mínima de Santa Rosa
temperaturas_minimas = santa_rosa[colunas_tmin].iloc[0]

# Identificando o mês de menor temperatura
mes_menor = temperaturas_minimas.idxmin()
temperatura_menor = temperaturas_minimas.min()

# Traduzindo a abreviação do mês
meses = {
    'jan': 'Janeiro', 'fev': 'Fevereiro',
    'mar': 'Março', 'abr': 'Abril',
    'mai': 'Maio', 'jun': 'Junho',
    'jul': 'Julho', 'ago': 'Agosto',
    'set': 'Setembro', 'out': 'Outubro',
    'nov': 'Novembro', 'dez': 'Dezembro'
}

mes = mes_menor.replace('tmin_', '')

print(f"Mês de menor temperatura mínima: {meses[mes]}")
print(f"Temperatura mínima: {temperatura_menor:.2f} °C")
```

```python
# Selecionando as colunas de déficit hídrico
colunas_def = [
    'def_jan', 'def_fev', 'def_mar', 'def_abr',
    'def_mai', 'def_jun', 'def_jul', 'def_ago',
    'def_set', 'def_out', 'def_nov', 'def_dez'
]

# Obtendo os valores de déficit hídrico de Santa Rosa
deficit_hidrico = santa_rosa[colunas_def].iloc[0]

# Identificando o mês de maior déficit
mes_maior_deficit = deficit_hidrico.idxmax()
valor_maior_deficit = deficit_hidrico.max()

# Traduzindo a abreviação do mês
mes = mes_maior_deficit.replace('def_', '')

print(f"Mês de maior déficit hídrico: {meses[mes]}")
print(f"Déficit hídrico: {valor_maior_deficit:.2f}")
```

```python
# Definindo os meses de cada estação
estacoes = {
    'Verão': [
        'temp_media_dez', 'temp_media_jan', 'temp_media_fev'
    ],
    'Outono': [
        'temp_media_mar', 'temp_media_abr', 'temp_media_mai'
    ],
    'Inverno': [
        'temp_media_jun', 'temp_media_jul', 'temp_media_ago'
    ],
    'Primavera': [
        'temp_media_set', 'temp_media_out', 'temp_media_nov'
    ]
}

# Calculando a média de temperatura de cada estação
medias_estacoes = {}

for estacao, colunas in estacoes.items():
    medias_estacoes[estacao] = (
        santa_rosa[colunas].mean(axis=1).iloc[0]
    )

# Organizando os resultados em uma tabela
resultado_estacoes = pd.DataFrame(
    medias_estacoes.items(),
    columns=['Estação', 'Temperatura média (°C)']
)

display(resultado_estacoes.round(2))
```
