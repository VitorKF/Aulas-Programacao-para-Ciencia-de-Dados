```python
# Definindo uma classe Pet e seus atributos
class Pet:

    # método construtor
    def __init__(self, nome: str, idade: int, peso: float, cor: str, tipo: str, __cadastrado=False):
        self.nome = nome
        self.idade = idade
        self.peso = peso
        self.cor = cor
        self.__tipo = tipo

        # Atributo privado
        self.__cadastrado = __cadastrado

    def setTipo(self, tipo: str):
        self.__tipo = tipo

    def getTipo(self):
        return self.__tipo

    def setCor(self, cor: str):
        self.cor = cor

    def getCor(self):
        return self.cor
```

```python
# Definindo uma classe Pet e seus atributos
# Já estamos usando Abstração ao definir a classe

class Passaro(Pet):

    # método construtor
    def __init__(self, nome: str, idade: int, peso: float, cor: str, bico: str):
        super().__init__(nome, idade, peso, cor, 'Passaro')
        self.bico = bico

    def setBico(self, bico: str):
        self.bico = bico

    def getBico(self):
        return self.bico

    def tipoComida(self):
        return (
            "Alpiste"
        )
```

```python
cardeal = Passaro(nome='Pardal', idade=2, peso=0.7, cor='Amarela', bico='Fino')

print(cardeal.nome)
print(cardeal.bico)
print(cardeal.getCor())
print(cardeal.getTipo())
print(cardeal.tipoComida())
```
